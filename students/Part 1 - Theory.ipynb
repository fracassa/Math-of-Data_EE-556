## 1. Multiclass Softmax Logistic Regression - 30 Points


We now model multiclass classification with classes $c \in \{1,\dots,C\}$. For each sample $(\mathbf{a}_i, b_i)$ with $\mathbf{a}_i \in \mathbb{R}^p$ and $b_i \in \{1,\dots,C\}$, let $\mathbf{X} = [\mathbf{x}_1,\dots,\mathbf{x}_C] \in \mathbb{R}^{p\times C}$ be the class weight matrix. The softmax model defines

$$
\mathbb{P}(b_i = c \mid \mathbf{a}_i) = \frac{\exp(\mathbf{a}_i^\top \mathbf{x}_c)}{\sum_{k=1}^C \exp(\mathbf{a}_i^\top \mathbf{x}_k)}.
$$

Assume i.i.d. samples $\{(\mathbf{a}_i,b_i)\}_{i=1}^n$. Our goal is to estimate $\mathbf{X}$ by maximum likelihood (and later with an $\ell_2$ regularizer).


__(a)__ (2 points) Show that the negative log-likelihood $f$ can be written as:

$$
\begin{aligned}
 f(\mathbf{X})
 &= - \log \mathbb{P}(b_1,\dots,b_n\mid \mathbf{a}_1,\dots,\mathbf{a}_n)\\
 &= \sum_{i=1}^n \left[ -\mathbf{a}_i^\top \mathbf{x}_{b_i} + \log \sum_{k=1}^C \exp(\mathbf{a}_i^\top \mathbf{x}_k) \right].
\end{aligned}
$$



**ANSWER:**

By independence of the samples, the conditional likelihood is

$$ 
P(b_1,\ldots,b_n\mid a_1,\ldots,a_n;X)
=
\prod_{i=1}^{n}P(b_i\mid a_i;X)
=
\prod_{i=1}^{n}
\frac{e^{a_i^\top x_{b_i}}}
{\sum_{k=1}^{C}e^{a_i^\top x_k}}.
$$

Taking the negative logarithm gives

$$
\begin{aligned}
f(X)
&=
-\log P(b_1,\ldots,b_n\mid a_1,\ldots,a_n;X)\\
&=
-\sum_{i=1}^{n}
\left[
\log\left(e^{a_i^\top x_{b_i}}\right)
-
\log\left(\sum_{k=1}^{C}e^{a_i^\top x_k}\right)
\right]\\
&=
\sum_{i=1}^{n}
\left[
-a_i^\top x_{b_i}
+
\log\left(\sum_{k=1}^{C}e^{a_i^\top x_k}\right)
\right].
\end{aligned}
$$


__(b)__ (3 points) Show that $\mathbf{u} \mapsto \log\!\left(\sum_{k=1}^C e^{u_k}\right)$ is convex on $\mathbb{R}^C$. Then, show that $f(\mathbf{X})$ is convex.


Hint: use Jensen's inequality.

**ANSWER:**

Let

$$
h(u)=\log\left(\sum_{k=1}^{C}e^{u_k}\right),
\qquad u\in\mathbb R^C.
$$

To prove convexity, we need to show that

$$
h(\lambda u+(1-\lambda)v)
\le
\lambda h(u)+(1-\lambda)h(v)
$$

for all $ u,v\in\mathbb R^C $ and $\lambda\in[0,1]$.

For $0<\lambda<1$, write

$$
h(\lambda u+(1-\lambda)v)
=
\log\left(
\sum_{k=1}^{C}e^{\lambda u_k}e^{(1-\lambda)v_k}
\right).
$$

Hölder's inequality states that

$$
\sum_k |A_kB_k|
\le
\left(\sum_k |A_k|^r\right)^{1/r}
\left(\sum_k |B_k|^s\right)^{1/s},
\qquad
\frac1r+\frac1s=1.
$$

Choose

$$
A_k=e^{\lambda u_k},
\qquad
B_k=e^{(1-\lambda)v_k},
\qquad
r=\frac1\lambda,
\qquad
s=\frac1{1-\lambda}.
$$

Then

$$
\sum_{k=1}^{C}e^{\lambda u_k}e^{(1-\lambda)v_k}
\le
\left(\sum_{k=1}^{C}e^{u_k}\right)^\lambda
\left(\sum_{k=1}^{C}e^{v_k}\right)^{1-\lambda}.
$$

Since the logarithm is increasing, taking logarithms yields

$$
h(\lambda u+(1-\lambda)v)
\le
\lambda\log\left(\sum_{k=1}^{C}e^{u_k}\right)
+
(1-\lambda)\log\left(\sum_{k=1}^{C}e^{v_k}\right).
$$

This is the required convexity inequality. The cases
\(\lambda=0\) and \(\lambda=1\) hold with equality.

Finally, we can write

$$
f(X)=
\sum_{i=1}^{n}
\left[-a_i^\top x_{b_i}+h(X^\top a_i)\right].
$$

The map $X\mapsto X^\top a_i$ is linear, so
$X\mapsto h(X^\top a_i)$ is convex. The term
$-a_i^\top x_{b_i}$ is linear. Therefore each summand is
convex, and hence $f$ is convex.


Since the negative log-likelihood $f$ is convex, every local minimizer is a global minimizer. The maximum likelihood estimation problem is
$$
\mathbf{X}^\star_{ML} = \arg\min_{\mathbf{X} \in \mathbb{R}^{p\times C}} f(\mathbf{X})
$$

But does $f$ always attain its infimum? The following three questions examine this issue.


__(c)__ (1 point) Explain the difference between infima and minima. Give an example of a convex function on $\mathbb{R}$ that does not attain its infimum.


**ANSWER:**

Let $\varphi:\mathcal D\to\mathbb R$ with $\mathcal D\neq\emptyset$. Its **infimum** is the
greatest lower bound of its values:

$$
\inf_{x\in\mathcal D}\varphi(x)
=
\sup\bigl\{\ell\in\mathbb R:\ \varphi(x)\ge\ell\ \ \forall x\in\mathcal D\bigr\}
\in\mathbb R\cup\{-\infty\},
$$

with the convention $\sup\emptyset=-\infty$, so $\inf\varphi=-\infty$ when $\varphi$ is
unbounded below. The infimum always exists. 

A **minimum** exists only if the infimum is attained:

$$
\min_{x\in\mathcal D}\varphi(x)\ \text{exists}
\iff
\exists\,x^\star\in\mathcal D:\ \varphi(x^\star)=\inf_{x\in\mathcal D}\varphi(x),
$$

and in that case $\min\varphi=\varphi(x^\star)$. Every minimum is an infimum, but the converse
fails: the values $\varphi(x)$ can get arbitrarily close to $\varphi^\star$ without ever reaching it.

For example, consider
$$
g(t)=e^t,\qquad t\in\mathbb R.
$$

This function is convex because $g''(t)=e^t>0$. Moreover,

$$
g(t)>0
\quad\text{for every finite }t,
\qquad
\lim_{t\to-\infty}g(t)=0.
$$

Thus $\inf_{t\in\mathbb R}g(t)=0$, but there is no $t\in\mathbb R$
such that $g(t)=0$. 

The infimum is therefore not attained.

__(d)__ (2 points) Assume there exists $\mathbf{X}_0 \in \mathbb{R}^{p\times C}$ such that for all $i$,
$$
\mathbf{a}_i^\top \mathbf{x}_{0, b_i} - \max_{k \neq b_i} \mathbf{a}_i^\top \mathbf{x}_{0,k} > 0.
$$
This is called one-versus-all complete separation in multiclass settings. Give a geometric interpretation (e.g., for $p=2$) and explain why the name is appropriate.


**ANSWER:**

The separation assumption is equivalent to

$$
a_i^\top(x_{0,b_i}-x_{0,k})>0
\qquad
\text{for every }i\text{ and every }k\ne b_i.
$$

For two classes $c$ and $k$, the equation

$$
a^\top(x_{0,c}-x_{0,k})=0
$$

defines the boundary where their scores are equal, and for $p=2$,
this is a line through the origin. On one side, class $c$ has
the higher score, on the other, class $k$ does.

The region where class $c$ has a strictly larger score than all
other classes is the intersection of the half-spaces

$$
\left\{
a\in\mathbb R^p:
a^\top(x_{0,c}-x_{0,k})>0
\right\},
\qquad k\ne c.
$$

Complete separation means that every training sample lies strictly
inside the region corresponding to its correct class. Thus all
training samples are classified correctly, with a strictly positive
score margin against every competing class. 

This explains the name:
the separation holds for every sample and compares its correct
class against all the others.


Complete separation allows perfect classification of the training data. However, as you will show next, it prevents the existence of a finite maximum likelihood estimate.


__(e)__ (3 points) In a one-versus-all complete separation setting (as in (d)), prove that $f$ does not attain its minimum. Hint: consider $f(\alpha \mathbf{X}_0)$ as $\alpha \to +\infty$ and compare it to $f(\mathbf{X}_0)$.


**ANSWER:**

Let $X_0$ satisfy the complete separation assumption, and define

$$
m_{i,k}
=
a_i^\top(x_{0,b_i}-x_{0,k})>0,
\qquad k\ne b_i.
$$

Consider $X=\alpha X_0$, with $\alpha>0$; factoring out the exponential
of the correct-class score inside each logarithm gives

$$
\begin{aligned}
f(\alpha X_0)
&=
\sum_{i=1}^{n}
\left[
-\alpha a_i^\top x_{0,b_i}
+
\log\left(\sum_{k=1}^{C}e^{\alpha a_i^\top x_{0,k}}\right)
\right]\\
&=
\sum_{i=1}^{n}
\log\left(
1+\sum_{k\ne b_i}
e^{\alpha a_i^\top(x_{0,k}-x_{0,b_i})}
\right)\\
&=
\sum_{i=1}^{n}
\log\left(
1+\sum_{k\ne b_i}e^{-\alpha m_{i,k}}
\right).
\end{aligned}
$$

Since every $m_{i,k}>0$, each exponential tends to zero as
$\alpha\to+\infty$. Hence

$$
\lim_{\alpha\to+\infty}f(\alpha X_0)=0.
$$

In particular, for $\alpha>1$, every exponential is strictly smaller
than at $\alpha=1$, so $f(\alpha X_0)<f(X_0)$.

To conclude, we must also check whether the limiting value can
be attained. For any finite $X$, assuming $C\ge2$, all
exponentials in the softmax are strictly positive. Consequently,

$$
0<P(b_i\mid a_i;X)<1,
$$

and therefore, for $n\ge1$

$$
f(X)
=
-\sum_{i=1}^{n}\log P(b_i\mid a_i;X)>0.
$$

Thus $f$ is bounded below by zero and approaches zero along
$\alpha X_0$, but never equals zero at a finite matrix:

$$
\inf_{X\in\mathbb R^{p\times C}}f(X)=0,
\qquad
f(X)>0\text{ for every finite }X.
$$

The infimum is not attained, so no finite maximum-likelihood
estimate exists.


We resolve this issue by adding a regularizer. Consider the regularized function

$$
 f_\mu(\mathbf{X}) = f(\mathbf{X}) + \frac{\mu}{2} \|\mathbf{X}\|_F^2, \quad \mu > 0.
$$

__(f)__ (2 points) Show that the gradient with respect to $\mathbf{X}$ of $f_\mu$ can be expressed as
$$
 \nabla_{\mathbf{X}} f_\mu(\mathbf{X}) = \sum_{i=1}^n \mathbf{a}_i \big( \mathbf{p}_i - \mathbf{e}_{b_i} \big)^\top + \mu \mathbf{X},
$$
where $\mathbf{e}_{b_i} \in \mathbb{R}^C$ is the [one-hot vector](https://en.wikipedia.org/wiki/One-hot) for class $b_i$, $\mathbf{p}_i \in \mathbb{R}^C$ has entries $p_{i,c} = \mathbb{P}(b_i=c\mid \mathbf{a}_i)$ under the softmax model, and $\mathbf{a}_i(\mathbf{p}_i - \mathbf{e}_{b_i})^\top\in\mathbb{R}^{p\times C}$ denotes the outer product.



**ANSWER:**

We compute the gradient by separating the linear term, the
log-sum-exp term and the regularization.

For each column of the weight matrix, the linear term gives

$$
\nabla_{x_c}(-a_i^\top x_{b_i})
=
-\delta_{c,b_i}a_i.
$$

For the log-sum-exp term, applying the chain rule gives

$$
\begin{aligned}
\nabla_{x_c}
\log\left(\sum_{k=1}^{C}e^{a_i^\top x_k}\right)
&=
\frac{
\sum_{k=1}^{C}
e^{a_i^\top x_k}\nabla_{x_c}(a_i^\top x_k)
}{
\sum_{k=1}^{C}e^{a_i^\top x_k}
}\\
&=
\frac{
\sum_{k=1}^{C}e^{a_i^\top x_k}\delta_{kc}a_i
}{
\sum_{k=1}^{C}e^{a_i^\top x_k}
}\\
&=
\frac{e^{a_i^\top x_c}}
{\sum_{k=1}^{C}e^{a_i^\top x_k}}a_i
=
p_{i,c}a_i.
\end{aligned}
$$

Since the squared Frobenius norm is the sum of the squared norms
of the columns, the regularization contributes as

$$
\nabla_{x_c}\left(\frac{\mu}{2}\|X\|_F^2\right)
=
\nabla_{x_c}\left(
\frac{\mu}{2}\sum_{k=1}^{C}x_k^\top x_k
\right)
=
\mu x_c.
$$

Therefore,

$$
\nabla_{x_c}f_\mu(X)
=
\sum_{i=1}^{n}a_i(p_{i,c}-\delta_{c,b_i})
+
\mu x_c.
$$

The components of the one-hot vector satisfy

$$
[e_{b_i}]_c=\delta_{c,b_i}.
$$

Thus the column indexed by c is the corresponding column of
the outer product below; by collecting all columns we obtain:

$$
\nabla_X f_\mu(X)
=
\sum_{i=1}^{n}a_i(p_i-e_{b_i})^\top+\mu X.
$$

__(g)__ (4 points) Use row-wise vectorization of $\mathbf{X}$, obtained by stacking its rows one after another into a vector in $\mathbb{R}^{pC}$. Show that the Hessian of $f_\mu$ in these coordinates can be written as
<a id="eq-softmax-hessian"></a>

$$
 \nabla^2 f_\mu(\mathbf{X}) = \sum_{i=1}^n (\mathbf{a}_i\mathbf{a}_i^\top) \otimes \big( \operatorname{Diag}(\mathbf{p}_i) - \mathbf{p}_i\mathbf{p}_i^\top \big) + \mu \mathbf{I}_{pC},\tag{1}
$$
where $\otimes$ is the Kronecker product, and $\operatorname{Diag}(\mathbf{p}_i) - \mathbf{p}_i\mathbf{p}_i^\top$ is the softmax Jacobian, which is positive semidefinite.


**ANSWER:**

From part f, the gradient entry indexed by r and c is

$$
\frac{\partial f_\mu}{\partial X_{r,c}}
=
\sum_{i=1}^{n}a_{i,r}(p_{i,c}-\delta_{c,b_i})
+
\mu X_{r,c}.
$$

We differentiate this expression with respect to another entry,
indexed by s and d. First, since

$$
\frac{\partial(a_i^\top x_k)}{\partial X_{s,d}}
=
a_{i,s}\delta_{kd},
$$

the derivatives of the numerator and denominator of the softmax are

$$
\frac{\partial e^{a_i^\top x_c}}{\partial X_{s,d}}
=
a_{i,s}\delta_{cd}e^{a_i^\top x_c},
\qquad
\frac{\partial}{\partial X_{s,d}}
\sum_{k=1}^{C}e^{a_i^\top x_k}
=
a_{i,s}e^{a_i^\top x_d}.
$$

By the quotient rule,

$$
\begin{aligned}
\frac{\partial p_{i,c}}{\partial X_{s,d}}
&=
a_{i,s}
\frac{
\delta_{cd}e^{a_i^\top x_c}
\sum_{k=1}^{C}e^{a_i^\top x_k}
-
e^{a_i^\top x_c}e^{a_i^\top x_d}
}{
\left(\sum_{k=1}^{C}e^{a_i^\top x_k}\right)^2
}\\
&=
a_{i,s}
\left(p_{i,c}\delta_{cd}-p_{i,c}p_{i,d}\right).
\end{aligned}
$$

The features and labels are fixed. Hence, differentiating the
gradient gives

$$
\frac{\partial}{\partial X_{s,d}}
\left(\frac{\partial f_\mu}{\partial X_{r,c}}\right)
=
\sum_{i=1}^{n}
a_{i,r}a_{i,s}
\left(p_{i,c}\delta_{cd}-p_{i,c}p_{i,d}\right)
+
\mu\delta_{rs}\delta_{cd}.
$$

The expression involving the probabilities is an entry of the
softmax Jacobian:

$$
\left[\operatorname{Diag}(p_i)-p_ip_i^\top\right]_{c,d}
=
p_{i,c}\delta_{cd}-p_{i,c}p_{i,d}.
$$

With row-wise vectorization, the entries of each row of the weight
matrix are stacked consecutively. The Hessian therefore has
p by p blocks, each of size C by C. The block indexed by r and s is

$$
\sum_{i=1}^{n}
a_{i,r}a_{i,s}
\left(\operatorname{Diag}(p_i)-p_ip_i^\top\right)
+
\mu\delta_{rs}I_C.
$$

In the Kronecker product, each entry of the first factor multiplies
the entire second factor. Since

$$
[a_i a_i^\top]_{r,s}=a_{i,r}a_{i,s},
$$

these blocks give

$$
\nabla^2 f_\mu(X)
=
\sum_{i=1}^{n}
(a_i a_i^\top)\otimes
\left(\operatorname{Diag}(p_i)-p_ip_i^\top\right)
+
\mu I_{pC}.
$$

To verify that the softmax Jacobian is positive semidefinite,
take an arbitrary vector and define its weighted mean:

$$
z\in\mathbb R^C,
\qquad
\bar z_i=\sum_{c=1}^{C}p_{i,c}z_c.
$$

Using that the probabilities sum to one,

$$
\begin{aligned}
z^\top\left(\operatorname{Diag}(p_i)-p_ip_i^\top\right)z
&=
\sum_{c=1}^{C}p_{i,c}z_c^2
-
\left(\sum_{c=1}^{C}p_{i,c}z_c\right)^2\\
&=
\sum_{c=1}^{C}p_{i,c}(z_c-\bar z_i)^2
\ge0.
\end{aligned}
$$

Thus the softmax Jacobian is positive semidefinite.

__(h)__ (2 points) Show that $f_\mu$ is $\mu$-strongly convex.


**ANSWER**

Recall that $h$ is $\mu$-strongly convex if and only if $h - \frac{\mu}{2}\|\cdot\|^2$ is convex. Here

$$
f_\mu(\mathbf{X}) - \frac{\mu}{2}\|\mathbf{X}\|_F^2 = f(\mathbf{X}),
$$

which is convex by (b). Hence $f_\mu$ is $\mu$-strongly convex. $\blacksquare$

*Equivalently, using the Hessian (1).* A twice-differentiable function is $\mu$-strongly convex if and only if $\nabla^2 f_\mu(\mathbf{X})\succeq\mu\mathbf{I}$ for all $\mathbf{X}$. The eigenvalues of a Kronecker product of symmetric matrices are the products of their eigenvalues (see (j-2)). Since $\mathbf{a}_i\mathbf{a}_i^\top\succeq 0$ and $\mathbf{S}_i\succeq 0$ (see (g)), we get $(\mathbf{a}_i\mathbf{a}_i^\top)\otimes\mathbf{S}_i\succeq 0$. Therefore $\nabla^2 f_\mu(\mathbf{X}) - \mu\mathbf{I}_{pC} = \sum_i(\mathbf{a}_i\mathbf{a}_i^\top)\otimes\mathbf{S}_i\succeq 0$.


__(i)__ (2 points) Is it possible for a strongly convex function to not attain its minimum? Justify your reasoning (you may assume the domain is $\mathbb{R}^{p\times C}$).


**ANSWER**

**No.** A finite-valued $\mu$-strongly convex function $h$ on $\mathbb{R}^{p\times C}$ always attains its minimum, and the minimizer is unique.

*Existence.* Fix any $\mathbf{X}_0$. The function $\varphi := h - \frac{\mu}{2}\|\cdot\|_F^2$ is convex and finite on $\mathbb{R}^{p\times C}$, hence continuous and subdifferentiable. So there is $\mathbf{s}$ with $\varphi(\mathbf{Y})\ge\varphi(\mathbf{X}_0) + \langle\mathbf{s},\mathbf{Y}-\mathbf{X}_0\rangle$ for all $\mathbf{Y}$; for our differentiable $f_\mu$, simply $\mathbf{s} = \nabla f(\mathbf{X}_0)$. Add $\frac{\mu}{2}\|\mathbf{Y}\|_F^2 = \frac{\mu}{2}\|\mathbf{X}_0\|_F^2 + \mu\langle\mathbf{X}_0,\mathbf{Y}-\mathbf{X}_0\rangle + \frac{\mu}{2}\|\mathbf{Y}-\mathbf{X}_0\|_F^2$ to both sides, set $\mathbf{g} := \mathbf{s}+\mu\mathbf{X}_0$, and apply Cauchy–Schwarz:

$$
h(\mathbf{Y}) \ge h(\mathbf{X}_0) + \langle\mathbf{g},\mathbf{Y}-\mathbf{X}_0\rangle + \frac{\mu}{2}\|\mathbf{Y}-\mathbf{X}_0\|_F^2
\ge h(\mathbf{X}_0) + \|\mathbf{Y}-\mathbf{X}_0\|_F\Big(\frac{\mu}{2}\|\mathbf{Y}-\mathbf{X}_0\|_F - \|\mathbf{g}\|_F\Big).
$$

The right-hand side exceeds $h(\mathbf{X}_0)$ whenever $\|\mathbf{Y}-\mathbf{X}_0\|_F > 2\|\mathbf{g}\|_F/\mu$. Hence the sublevel set $\mathcal{S} := \{\mathbf{Y}: h(\mathbf{Y})\le h(\mathbf{X}_0)\}$ is bounded. It is also closed, by continuity, and non-empty, since $\mathbf{X}_0\in\mathcal{S}$; so it is compact. By Weierstrass' theorem, $h$ attains its minimum over $\mathcal{S}$ at some $\mathbf{X}^\star$. Outside $\mathcal{S}$ we have $h > h(\mathbf{X}_0)\ge h(\mathbf{X}^\star)$, so $\mathbf{X}^\star$ is a global minimizer.

*Uniqueness.* Suppose $\mathbf{X}_1\neq\mathbf{X}_2$ were two minimizers with value $h^\star$. Strong convexity at the midpoint would give $h\big(\tfrac{\mathbf{X}_1+\mathbf{X}_2}{2}\big)\le\tfrac12 h(\mathbf{X}_1)+\tfrac12 h(\mathbf{X}_2) - \tfrac{\mu}{8}\|\mathbf{X}_1-\mathbf{X}_2\|_F^2 < h^\star$, a contradiction. $\blacksquare$

So the infimum cannot "escape to infinity" as in (e): the quadratic growth forced by strong convexity confines every minimizing sequence to a compact set. In particular, for every $\mu>0$, $f_\mu$ has a unique minimizer, even under complete separation. The assumption on the domain matters: for example, $h(x) = x^2$ on the open interval $(1,2)$ is strongly convex but has no minimum.

  \leftarrow &  \mathbf{a}_n^\top & \rightarrow \\
 \end{bmatrix}.
$$

__(j-1)__ Show that $\lambda_{\max}(\mathbf{a}_i\mathbf{a}_i^T) = \left\| \mathbf{a}_i\right\|_2^2$, where $\lambda_{\max}(\cdot)$ denotes the largest eigenvalue.


**ANSWER**

The matrix $\mathbf{a}_i\mathbf{a}_i^\top$ is symmetric, so by the Rayleigh-quotient characterization

$$
\lambda_{\max}(\mathbf{a}_i\mathbf{a}_i^\top) = \max_{\|\mathbf{v}\|_2=1}\mathbf{v}^\top\mathbf{a}_i\mathbf{a}_i^\top\mathbf{v} = \max_{\|\mathbf{v}\|_2=1}\big(\mathbf{a}_i^\top\mathbf{v}\big)^2 .
$$

* *Upper bound.* By Cauchy–Schwarz, $(\mathbf{a}_i^\top\mathbf{v})^2\le\|\mathbf{a}_i\|_2^2\,\|\mathbf{v}\|_2^2 = \|\mathbf{a}_i\|_2^2$ for every unit vector $\mathbf{v}$.
* *The bound is attained.* If $\mathbf{a}_i\neq\mathbf{0}$, take $\mathbf{v} = \mathbf{a}_i/\|\mathbf{a}_i\|_2$; then $(\mathbf{a}_i^\top\mathbf{v})^2 = \|\mathbf{a}_i\|_2^2$. Equivalently, $(\mathbf{a}_i\mathbf{a}_i^\top)\,\mathbf{a}_i = \|\mathbf{a}_i\|_2^2\,\mathbf{a}_i$, so $\mathbf{a}_i$ is an eigenvector with eigenvalue $\|\mathbf{a}_i\|_2^2$. If $\mathbf{a}_i = \mathbf{0}$, both sides are $0$.

Hence $\lambda_{\max}(\mathbf{a}_i\mathbf{a}_i^\top) = \|\mathbf{a}_i\|_2^2$. $\blacksquare$ (The other $p-1$ eigenvalues are $0$, with eigenvectors orthogonal to $\mathbf{a}_i$.)

__(j-2)__ Using [(1)](#eq-softmax-hessian), show that $\lambda_{\max}(\nabla^2 f_\mu(\mathbf{X})) \leq \sum_{i=1}^{n} \|\mathbf{a}_i\|_2^2 + \mu$.

**ANSWER**

We use two properties of symmetric matrices.

1. **Subadditivity.** $\lambda_{\max}(\mathbf{M}+\mathbf{N})\le\lambda_{\max}(\mathbf{M})+\lambda_{\max}(\mathbf{N})$, because $\max_{\|\mathbf{v}\|=1}\mathbf{v}^\top(\mathbf{M}+\mathbf{N})\mathbf{v}\le\max_{\|\mathbf{v}\|=1}\mathbf{v}^\top\mathbf{M}\mathbf{v}+\max_{\|\mathbf{v}\|=1}\mathbf{v}^\top\mathbf{N}\mathbf{v}$.
2. **Spectrum of a Kronecker product.** Let $\mathbf{M}\mathbf{u}_j = \lambda_j\mathbf{u}_j$ and $\mathbf{N}\mathbf{w}_k = \nu_k\mathbf{w}_k$, with orthonormal eigenbases $\{\mathbf{u}_j\}$ and $\{\mathbf{w}_k\}$. Then $(\mathbf{M}\otimes\mathbf{N})(\mathbf{u}_j\otimes\mathbf{w}_k) = \mathbf{M}\mathbf{u}_j\otimes\mathbf{N}\mathbf{w}_k = \lambda_j\nu_k\,(\mathbf{u}_j\otimes\mathbf{w}_k)$. The vectors $\mathbf{u}_j\otimes\mathbf{w}_k$ form an orthonormal basis, because $\langle\mathbf{u}_j\otimes\mathbf{w}_k,\mathbf{u}_{j'}\otimes\mathbf{w}_{k'}\rangle = \langle\mathbf{u}_j,\mathbf{u}_{j'}\rangle\langle\mathbf{w}_k,\mathbf{w}_{k'}\rangle$. So the eigenvalues of $\mathbf{M}\otimes\mathbf{N}$ are exactly the products $\lambda_j\nu_k$. In particular, for positive semidefinite $\mathbf{M},\mathbf{N}$ we get $\lambda_{\max}(\mathbf{M}\otimes\mathbf{N}) = \lambda_{\max}(\mathbf{M})\,\lambda_{\max}(\mathbf{N})$.

Both $\mathbf{a}_i\mathbf{a}_i^\top$ and $\mathbf{S}_i = \operatorname{Diag}(\mathbf{p}_i)-\mathbf{p}_i\mathbf{p}_i^\top$ are positive semidefinite. Also $\lambda_{\max}(\mathbf{a}_i\mathbf{a}_i^\top) = \|\mathbf{a}_i\|_2^2$ by (j-1), and $\lambda_{\max}(\mathbf{S}_i) = \|\mathbf{S}_i\|_2\le\frac12\le 1$. Applying properties 1 and 2 to (1):

$$
\lambda_{\max}\big(\nabla^2 f_\mu(\mathbf{X})\big)
\le\sum_{i=1}^n\lambda_{\max}\big((\mathbf{a}_i\mathbf{a}_i^\top)\otimes\mathbf{S}_i\big) + \lambda_{\max}(\mu\mathbf{I}_{pC})
= \sum_{i=1}^n\|\mathbf{a}_i\|_2^2\,\lambda_{\max}(\mathbf{S}_i) + \mu
\le \sum_{i=1}^n\|\mathbf{a}_i\|_2^2 + \mu. \qquad\blacksquare
$$

The looser bound $\lambda_{\max}(\mathbf{S}_i)\le 1$ also follows directly from (g): $\mathbf{v}^\top\mathbf{S}_i\mathbf{v} = \operatorname{Var}_{c\sim\mathbf{p}_i}(v_c)\le\sum_c p_{i,c}v_c^2\le\max_c v_c^2\le\|\mathbf{v}\|_2^2$. With the sharper constant $\frac12$, we even get $\lambda_{\max}\big(\nabla^2 f_\mu(\mathbf{X})\big)\le\frac12\sum_i\|\mathbf{a}_i\|_2^2+\mu$.

__(j-3)__ Conclude that $f_\mu$ is $L$-smooth for $L = \|\mathbf{A}\|_F^2 + \mu$.


**ANSWER**

First, $\sum_{i=1}^n\|\mathbf{a}_i\|_2^2 = \sum_{i=1}^n\sum_{j=1}^p a_{ij}^2 = \|\mathbf{A}\|_F^2$, so (j-2) reads $\lambda_{\max}\big(\nabla^2 f_\mu(\mathbf{X})\big)\le L$ for all $\mathbf{X}$. Since $\nabla^2 f_\mu(\mathbf{X})\succeq\mu\mathbf{I}\succ 0$, its spectral norm equals its largest eigenvalue. Hence $\|\nabla^2 f_\mu(\mathbf{X})\|_2\le L$ everywhere.

The function $f_\mu$ is twice continuously differentiable, because $\operatorname{lse}$ and the softmax are smooth. Row-wise vectorization is an isometry, $\|\mathbf{X}\|_F = \|\operatorname{vec}_r(\mathbf{X})\|_2$, so let $\mathbf{x}=\operatorname{vec}_r(\mathbf{X})$ and $\mathbf{y}=\operatorname{vec}_r(\mathbf{Y})$. By the fundamental theorem of calculus along the segment from $\mathbf{y}$ to $\mathbf{x}$,

$$
\nabla f_\mu(\mathbf{x})-\nabla f_\mu(\mathbf{y}) = \int_0^1\nabla^2 f_\mu\big(\mathbf{y}+t(\mathbf{x}-\mathbf{y})\big)\,(\mathbf{x}-\mathbf{y})\,dt .
$$

Therefore

$$
\big\|\nabla f_\mu(\mathbf{X})-\nabla f_\mu(\mathbf{Y})\big\|_F \le \int_0^1\big\|\nabla^2 f_\mu\big(\mathbf{y}+t(\mathbf{x}-\mathbf{y})\big)\big\|_2\,\|\mathbf{x}-\mathbf{y}\|_2\,dt \le L\,\|\mathbf{X}-\mathbf{Y}\|_F .
$$

So $\nabla f_\mu$ is $L$-Lipschitz with respect to the Frobenius norm, i.e. $f_\mu$ is $L$-smooth with $L = \|\mathbf{A}\|_F^2+\mu$. $\blacksquare$


__(l)__ (3 points) KL divergence and NLL. Let $q_i=(q(c\mid\mathbf{a}_i))_{c=1}^C$ be the true conditional label distribution and $\mathbf{p}_i$ the model softmax distribution. Write $\mathrm{KL}(q_i\,\|\,\mathbf{p}_i)$ and show that minimizing its average over the samples is equivalent to minimizing the expected average negative log-likelihood, where the expectation is over labels drawn from $q$ with the features held fixed. Explain how $f(\mathbf{X})/n$ from (a) estimates this expected loss.

**ANSWER**

Write $q_{i,c} := q(c\mid\mathbf{a}_i)$, and let $p_{i,c}(\mathbf{X})$ be the softmax probabilities. These are always $>0$, so the KL divergence is finite; we use the convention $0\log 0 = 0$.

$$
\mathrm{KL}(q_i\,\|\,\mathbf{p}_i) = \sum_{c=1}^C q_{i,c}\log\frac{q_{i,c}}{p_{i,c}(\mathbf{X})}
= \underbrace{\sum_{c=1}^C q_{i,c}\log q_{i,c}}_{=\,-H(q_i)} \;+\; \underbrace{\sum_{c=1}^C q_{i,c}\big(-\log p_{i,c}(\mathbf{X})\big)}_{=\,\mathrm{CE}(q_i,\,\mathbf{p}_i)} .
$$

The entropy $H(q_i)$ does not depend on $\mathbf{X}$, so

$$
\frac1n\sum_{i=1}^n\mathrm{KL}\big(q_i\,\|\,\mathbf{p}_i(\mathbf{X})\big) = R(\mathbf{X}) - \frac1n\sum_{i=1}^n H(q_i),
\qquad R(\mathbf{X}) := \frac1n\sum_{i=1}^n \mathrm{CE}\big(q_i,\mathbf{p}_i(\mathbf{X})\big).
$$

**$R$ is the expected average NLL.** By (a), $f(\mathbf{X})/n = \frac1n\sum_{i} -\log p_{i,b_i}(\mathbf{X})$. Draw the labels independently, $b_i\sim q_i$, and keep the features fixed. By linearity of expectation,

$$
\mathbb{E}_{b\sim q}\Big[\frac{f(\mathbf{X})}{n}\Big] = \frac1n\sum_{i=1}^n\mathbb{E}_{b_i\sim q_i}\big[-\log p_{i,b_i}(\mathbf{X})\big] = \frac1n\sum_{i=1}^n\sum_{c=1}^C q_{i,c}\big(-\log p_{i,c}(\mathbf{X})\big) = R(\mathbf{X}).
$$

The average KL divergence and the expected average NLL therefore differ by a constant that does not depend on $\mathbf{X}$. They have the same minimizers, so minimizing one is equivalent to minimizing the other. Moreover, $\mathrm{KL}\ge 0$ with equality iff $\mathbf{p}_i = q_i$, so $R(\mathbf{X})\ge\frac1n\sum_i H(q_i)$, with equality iff the model reproduces the true conditional distributions exactly.

**$f(\mathbf{X})/n$ as an estimator.** We never observe $q_i$, only one label $b_i\sim q_i$ per sample. The observed loss $f(\mathbf{X})/n$ replaces each expectation $\mathbb{E}_{b\sim q_i}[-\log p_{i,b}(\mathbf{X})]$ by its one-sample Monte Carlo estimate $-\log p_{i,b_i}(\mathbf{X})$. By the display above, this estimate is **unbiased**: $\mathbb{E}[f(\mathbf{X})/n] = R(\mathbf{X})$ for every $\mathbf{X}$. Since the labels are independent, its variance is $\frac{1}{n^2}\sum_i\operatorname{Var}_{b_i\sim q_i}\big[-\log p_{i,b_i}(\mathbf{X})\big]$. This is $\mathcal{O}(1/n)$ when the per-sample variances are bounded, so $f(\mathbf{X})/n$ concentrates around $R(\mathbf{X})$ as $n$ grows (law of large numbers). Maximum likelihood therefore minimizes an empirical estimate of the average KL divergence between the true and the model conditional distributions, up to an additive constant. If the features are also drawn i.i.d., the same argument shows that $f(\mathbf{X})/n$ estimates the population risk $\mathbb{E}_{\mathbf{a}}\big[\mathrm{CE}\big(q(\cdot\mid\mathbf{a}),\,\mathbf{p}(\mathbf{a};\mathbf{X})\big)\big]$.

The unregularized maximum likelihood problem may have no finite minimizer. Adding a quadratic regularizer with $\mu>0$ gives a unique regularized estimate, defined by the smooth, strongly convex problem:
<a id="eq-regularized-estimator"></a>

$$
 \mathbf{X}^\star = \arg\min_{\mathbf{X} \in \mathbb{R}^{p\times C}} f(\mathbf{X}) + \frac{\mu}{2}\|\mathbf{X}\|_F^2.\tag{2}
$$


## Binary logistic regression (specialization for Part 2)

While this part analyzed the multiclass (softmax) setting, in the next exercise we will continue under the simplified two-class case.

Let labels be $b_i \in \{-1, +1\}$, features $\mathbf{a}_i \in \mathbb{R}^p$, and weight vector $\mathbf{x} \in \mathbb{R}^p$. Define the sigmoid
$$
\sigma(t) = \frac{1}{1+e^{-t}}.
$$
Model the conditional distribution as
$$
\mathbb{P}(b_i = j \mid \mathbf{a}_i) = \sigma\big(j\, \mathbf{a}_i^\top \mathbf{x}\big), \quad j \in \{-1,+1\}.
$$
The likelihood over i.i.d. samples $\{(\mathbf{a}_i, b_i)\}_{i=1}^n$ is
$$
\mathcal{L}(\mathbf{x}) = \prod_{i=1}^n \sigma\big(b_i\, \mathbf{a}_i^\top \mathbf{x}\big),
$$
so the negative log-likelihood is
$$
 f(\mathbf{x}) = -\log \mathcal{L}(\mathbf{x}) = \sum_{i=1}^n \log\big(1 + e^{-b_i\, \mathbf{a}_i^\top \mathbf{x}}\big).
$$

__(m)__ (2 points) Show that the gradient of the negative log-likelihood is the standard binary logistic regression gradient:
$$
\nabla f(\mathbf{x}) = \sum_{i=1}^n \big(-b_i\, \sigma(-b_i\, \mathbf{a}_i^\top \mathbf{x})\big)\, \mathbf{a}_i.
$$
(Hint: use the chain rule and $\sigma'(t) = \sigma(t)\big(1-\sigma(t)\big)$.)

We will use this binary formulation in Part 2 - First order methods.


**ANSWER:**

We see that $\log(1+e^{-t}) = -\log(1) + \log(1+e^{-t}) = -\log(1/(1+e^{-t})) = -\log\sigma(t)$ and $1-\sigma(t) = \frac{e^{-t}}{1+e^{-t}} = \frac{1}{e^{t}+1} = \sigma(-t)$. Using $\sigma'(t) = \sigma(t)\big(1-\sigma(t)\big)$,

$$
\frac{d}{dt}\big[-\log\sigma(t)\big] = -\frac{\sigma'(t)}{\sigma(t)} = -\big(1-\sigma(t)\big) = -\sigma(-t).
$$

Each term of $f$ is this scalar function composed with $t_i(\mathbf{x}) := b_i\,\mathbf{a}_i^\top\mathbf{x}$, whose gradient is $\nabla t_i(\mathbf{x}) = b_i\mathbf{a}_i$. By the chain rule,

$$
\nabla f(\mathbf{x}) = \sum_{i=1}^n\big(-\sigma(-b_i\,\mathbf{a}_i^\top\mathbf{x})\big)\,b_i\mathbf{a}_i = \sum_{i=1}^n\big(-b_i\,\sigma(-b_i\,\mathbf{a}_i^\top\mathbf{x})\big)\,\mathbf{a}_i. \qquad\blacksquare
$$
